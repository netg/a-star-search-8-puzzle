```python
# A* Search Algorithm - 8 Puzzle

initial_state = [2, 8, 3, 1, 6, 4, 7, 0, 5]
goal_state =    [1, 2, 3, 8, 0, 4, 7, 6, 5]

h = 0
g = 0

blank_index = initial_state.index(0)

# Calculate heuristic h(x)
for index, tile in enumerate(initial_state):
    if tile != goal_state[index] and tile != 0:
        h += 1

print("Initial heuristic:", h)

f = g + h

# Move UP
up_index = blank_index - 3

new_state = initial_state.copy()

temp = new_state[blank_index]
new_state[blank_index] = new_state[up_index]
new_state[up_index] = temp

new_g = g + 1

h2 = 0

for index, tile in enumerate(new_state):
    if tile != goal_state[index] and tile != 0:
        h2 += 1

new_f = new_g + h2

# Move LEFT
left_index = blank_index - 1

new_state_left = initial_state.copy()

temp_left = new_state_left[blank_index]
new_state_left[blank_index] = new_state_left[left_index]
new_state_left[left_index] = temp_left

h_left = 0

for index, tile in enumerate(new_state_left):
    if tile != goal_state[index] and tile != 0:
        h_left += 1

f_left = new_g + h_left

# Move RIGHT
right_index = blank_index + 1

new_state_right = initial_state.copy()

temp_right = new_state_right[blank_index]
new_state_right[blank_index] = new_state_right[right_index]
new_state_right[right_index] = temp_right

h_right = 0

for index, tile in enumerate(new_state_right):
    if tile != goal_state[index] and tile != 0:
        h_right += 1

f_right = new_g + h_right

# Find the minimum cost
best_f = min(new_f, f_left, f_right)

print("Minimum cost:", best_f)

# Select the best move
if new_f == best_f:
    print("Best move: UP")
    best_state = new_state.copy()

elif f_left == best_f:
    print("Best move: LEFT")
    best_state = new_state_left.copy()

elif f_right == best_f:
    print("Best move: RIGHT")
    best_state = new_state_right.copy()

# Display the selected state
print("Selected state:", best_state)

# Check if the goal state has been reached
if best_state == goal_state:
    print("Solution found!")
else:
    print("Searching continues...")
```
